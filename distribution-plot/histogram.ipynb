## Plotting univariate histogram

```python
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.histplot(data = tips , x='total_bill')
```

## Figure level function

```python
sns.displot(data = tips , x= 'total_bill' , kind = 'hist')
```

## Bins

```python
sns.displot(data = tips , x = 'total_bill' , kind = 'hist' , bins=20)
```

```python
## its is possible to visualise the distributio of a actegorical variable using the logic of a histogram
## discrete bins are automatically set for the categorical feature
```

```python
tips
```

```python
sns.displot(data = tips , x = 'day' , kind = 'hist' , hue = 'day')
```

```python
## hue
sns.displot(data = tips , x = 'total_bill' , kind = 'hist' , hue = 'sex')
```

```python
# element -> step
sns.displot(data = tips , x = 'total_bill' , kind = 'hist' , hue = 'sex' , element = 'step')
```

```python
data = sns.load_dataset('titanic')
data
```

```python

sns.displot(data = data , x = 'age' , kind = 'hist' , hue = 'sex')
```

```python
## facet using col and row 
sns.displot(data = data , x = 'age' , kind = 'hist' , col = 'sex')
```
