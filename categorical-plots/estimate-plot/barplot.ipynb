```python
## A bar plot (or bar chart) is a graphical representation used to display and compare the frequency, count, or other measures (like mean, sum) of different categories using rectangular bars. The length or height of each bar corresponds to the value it represents.
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
tips = sns.load_dataset('tips')
```

```python
## barplot use kiya jata h jb ek categorical column ka el numerical column ke saath koi aggregated value calculate krna chhate hai
```

```python
sns.barplot(data = tips , x = 'sex' , y = 'total_bill')
## the average total bill of mals is around 20 $ and that of female is 18 $
```

```python
sns.barplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker')
```

## Estimator

```python
import numpy as np
sns.barplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = np.std)
```

```python
sns.barplot(data = tips , x = 'sex' , y = 'total_bill' ,  estimator = np.min)
```

## Pointplot


```python
sns.barplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker')
```

```python
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker')
```
