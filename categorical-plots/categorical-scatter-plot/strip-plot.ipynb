```python
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
## one column has to be categorical and one numerical
## using axes level function
sns.stripplot(data = tips , x = 'sex' , y = 'total_bill')
```

```python
## using figure level function
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind= 'strip')
```

```python
## jitter
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind= 'strip',jitter=0.2,hue = 'smoker')
```
