```python
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
gap = px.data.gapminder()
gap.head()
```

```python
gap_india = gap[gap['country'] == 'India']
```

## Axes level function

```python
sns.lineplot(data = gap_india , x = 'year' , y = 'lifeExp' )
plt.plot()
```

## Figure level function

```python
sns.relplot(data = gap_india , x = 'year' , y = 'lifeExp' , kind = 'line')
plt.show()
```

## Hue

```python
temp_df = gap[gap['country'].isin(['India' , 'Pakistan' , 'China'])]
temp_df.head()
```

```python
sns.relplot(data = temp_df , x = 'year' , y = 'lifeExp' , hue = 'country' , kind='line')
```
